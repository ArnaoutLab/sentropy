```python
from sentropy import sentropy
import numpy as np
import pandas as pd

# say each dataset is a fruit basket
P1 = np.array([12, 3, 1, 1])             
P2 = np.array([2,  2, 4, 4])              
count1 = pd.DataFrame({"apples": P1, "oranges": P2, }, index = ['Granny Smith', 'McIntosh', 'navel', 'cara cara'])        # package the classes as P
count1
```

```python
count1.to_csv('count1.csv', index = False)
```
